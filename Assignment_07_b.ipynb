## Question 1: Visualizing the 2PL Item Response Function

The probability of a correct response under the two-parameter logistic model is:

$$
p_i(\theta)
=
P(Y_i=1 \mid \Theta=\theta)
=
\frac{1}{1+\exp\left[-a_i(\theta-b_i)\right]}.
$$

To examine the effects of the discrimination parameter \(a_i\) and the difficulty parameter \(b_i\), the following four item characteristic curves are plotted:

1. \(a=0.8,\ b=0\)
2. \(a=2.0,\ b=-1\)
3. \(a=2.0,\ b=0\)
4. \(a=2.0,\ b=1\)

The discrimination value \(a=2.0\) is paired with three different difficulty values, \(b=-1\), \(b=0\), and \(b=1\). This allows us to observe how changing \(b\) shifts the item response curve horizontally.

```python
import numpy as np
import plotly.graph_objects as go


def two_pl_probability(theta, a, b):
    """
    Compute the probability of a correct response under the 2PL model.

    Parameters
    ----------
    theta : array-like
        Latent ability values.
    a : float
        Discrimination parameter.
    b : float
        Difficulty parameter.

    Returns
    -------
    array-like
        Probability of a correct response.
    """
    return 1 / (1 + np.exp(-a * (theta - b)))


# Ability grid
theta = np.linspace(-4, 4, 1000)

# Two different discrimination values are used.
# The value a = 2.0 is paired with three difficulty values.
parameter_sets = [
    (0.8, 0.0),
    (2.0, -1.0),
    (2.0, 0.0),
    (2.0, 1.0)
]

fig = go.Figure()

for a, b in parameter_sets:
    probability = two_pl_probability(theta, a, b)

    fig.add_trace(
        go.Scatter(
            x=theta,
            y=probability,
            mode="lines",
            name=f"a = {a}, b = {b}"
        )
    )

# Add a reference line at probability 0.5
fig.add_hline(
    y=0.5,
    line_dash="dash",
    annotation_text="Probability = 0.5",
    annotation_position="bottom right"
)

fig.update_layout(
    title="2PL Item Response Curves",
    xaxis_title="Latent Ability (θ)",
    yaxis_title="Probability of a Correct Response",
    template="plotly_white",
    hovermode="x unified",
    legend_title="Item Parameters"
)

fig.update_xaxes(range=[-4, 4])
fig.update_yaxes(range=[0, 1])

fig.show()
```

### Interpretation

The difficulty parameter \(b_i\) controls the horizontal position of the item response curve.

At the point where

$$
\theta=b_i,
$$

the probability of a correct response is

$$
p_i(b_i)
=
\frac{1}{1+\exp(0)}
=
\frac{1}{2}.
$$

Therefore, \(b_i\) is the ability level at which the user has a 50% probability of answering the item correctly.

When \(b_i\) increases, the response curve shifts to the right. This means that the user requires a higher ability level to achieve the same probability of answering correctly. Therefore, an item with a larger \(b_i\) is more difficult.

When \(b_i\) decreases, the curve shifts to the left. This means that users with lower ability levels have a higher probability of answering correctly. Therefore, an item with a smaller \(b_i\) is easier.

For the curves with \(a=2.0\):

- When \(b=-1\), the probability is 0.5 at \(\theta=-1\).
- When \(b=0\), the probability is 0.5 at \(\theta=0\).
- When \(b=1\), the probability is 0.5 at \(\theta=1\).

Thus, changing \(b_i\) shifts the curve horizontally without changing its general logistic shape.

The discrimination parameter \(a_i\) controls the steepness of the curve. The curve with \(a=2.0\) is steeper than the curve with \(a=0.8\). A larger discrimination value means that the item distinguishes more strongly between users whose abilities are slightly below and slightly above the item difficulty. A smaller discrimination value produces a flatter curve and provides weaker separation between ability levels.

## Question 2: Sequential Likelihood Contribution

For the \(k\)-th item, the probability of a correct response under the 2PL model is

$$
p_k(\theta)
=
P(Y_k=1\mid\Theta=\theta)
=
\frac{1}{1+\exp\left[-a_k(\theta-b_k)\right]}.
$$

The observed response satisfies

$$
y_k\in\{0,1\}.
$$

Therefore, the likelihood contribution of one new response \(y_k\), conditional on the latent ability \(\theta\), is

$$
L(y_k\mid\theta)
=
[p_k(\theta)]^{y_k}
[1-p_k(\theta)]^{1-y_k}.
$$

This expression covers both possible outcomes.

If the response is correct, \(y_k=1\), then

$$
L(1\mid\theta)
=
p_k(\theta).
$$

If the response is incorrect, \(y_k=0\), then

$$
L(0\mid\theta)
=
1-p_k(\theta).
$$

Substituting the 2PL response probability, the single-response likelihood becomes

$$
L(y_k\mid\theta)
=
\left[
\frac{1}{1+\exp\left[-a_k(\theta-b_k)\right]}
\right]^{y_k}
\left[
1-
\frac{1}{1+\exp\left[-a_k(\theta-b_k)\right]}
\right]^{1-y_k}.
$$

Let the running response history up to step \(k\) be

$$
\mathbf{y}^{(k)}
=
(y_1,y_2,\ldots,y_k).
$$

Assuming that the responses are conditionally independent given \(\theta\), the joint likelihood of the complete running response history is

$$
L(\mathbf{y}^{(k)}\mid\theta)
=
\prod_{i=1}^{k}
[p_i(\theta)]^{y_i}
[1-p_i(\theta)]^{1-y_i}.
$$

Using the 2PL response probability explicitly,

$$
L(\mathbf{y}^{(k)}\mid\theta)
=
\prod_{i=1}^{k}
\left[
\frac{1}{1+\exp\left[-a_i(\theta-b_i)\right]}
\right]^{y_i}
\left[
1-
\frac{1}{1+\exp\left[-a_i(\theta-b_i)\right]}
\right]^{1-y_i}.
$$

Each response contributes evidence about the user's latent ability. A correct response gives greater likelihood to ability values for which \(p_i(\theta)\) is large, while an incorrect response gives greater likelihood to ability values for which \(p_i(\theta)\) is small.

## Question 3: Mathematical Formulation of the Running Update

Let the observed response history after \(k\) items be

$$
\mathbf{y}^{(k)}
=
(y_1,y_2,\ldots,y_k).
$$

At step \(k-1\), the current posterior density is

$$
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right).
$$

This posterior becomes the prior distribution for the next update.

For the new item \(k\), the 2PL response probability is

$$
p_k(\theta)
=
\frac{1}
{1+\exp\left[-a_k(\theta-b_k)\right]}.
$$

The likelihood contribution of the newly observed response \(y_k\) is

$$
L(y_k\mid\theta)
=
[p_k(\theta)]^{y_k}
[1-p_k(\theta)]^{1-y_k}.
$$

Using Bayes' theorem, the posterior density after observing the new response is proportional to the previous posterior multiplied by the new likelihood contribution:

$$
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k)}
\right)
\propto
L(y_k\mid\theta)
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right).
$$

Substituting the Bernoulli likelihood gives

$$
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k)}
\right)
\propto
[p_k(\theta)]^{y_k}
[1-p_k(\theta)]^{1-y_k}
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right).
$$

Substituting the 2PL probability explicitly,

$$
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k)}
\right)
\propto
\left[
\frac{1}
{1+\exp\left[-a_k(\theta-b_k)\right]}
\right]^{y_k}
\left[
1-
\frac{1}
{1+\exp\left[-a_k(\theta-b_k)\right]}
\right]^{1-y_k}
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right).
$$

The normalized posterior density is

$$
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k)}
\right)
=
\frac{
L(y_k\mid\theta)
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right)
}{
\displaystyle
\int_{-\infty}^{\infty}
L(y_k\mid s)
f_{\Theta\mid Y}
\left(
s\mid\mathbf{y}^{(k-1)}
\right)
\,ds
}.
$$

The denominator is the normalization constant that ensures

$$
\int_{-\infty}^{\infty}
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k)}
\right)
\,d\theta
=
1.
$$

Therefore, the sequential Bayesian update can be summarized as

$$
\text{new posterior}
\propto
\text{new likelihood}
\times
\text{previous posterior}.
$$

Initially, before any responses are observed, the distribution is

$$
f_{\Theta}^{(0)}(\theta)
=
\frac{1}{\sqrt{2\pi}}
\exp\left(-\frac{\theta^2}{2}\right),
$$

which corresponds to the prior

$$
\Theta\sim N(0,1).
$$

## Question 4: Dynamic Shifting

At step \(k\), suppose the user answers a highly difficult item correctly. This means

$$
y_k=1
$$

and the difficulty parameter \(b_k\) is large.

For a correct response, the likelihood contribution is

$$
L(y_k=1\mid\theta)
=
p_k(\theta),
$$

where

$$
p_k(\theta)
=
\frac{1}{1+\exp\left[-a_k(\theta-b_k)\right]}.
$$

The sequential posterior update is

$$
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k)}
\right)
\propto
p_k(\theta)
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right).
$$

For a highly difficult item, \(b_k\) is large. Users with low ability values have a small probability of answering the item correctly because, when

$$
\theta<b_k,
$$

we have

$$
p_k(\theta)<0.5.
$$

In contrast, users with ability values near or above the item difficulty have a much larger probability of answering correctly.

Therefore, multiplying the previous posterior by \(p_k(\theta)\) reduces the posterior density at low values of \(\theta\) and gives relatively greater weight to higher values of \(\theta\).

As a result, the peak, or mode, of the posterior distribution generally shifts to the right toward larger ability values.

Mathematically, the new mode is

$$
\hat{\theta}_{MAP}^{(k)}
=
\operatorname*{arg\,max}_{\theta}
\left\{
p_k(\theta)
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right)
\right\}.
$$

Taking logarithms gives

$$
\hat{\theta}_{MAP}^{(k)}
=
\operatorname*{arg\,max}_{\theta}
\left\{
\log p_k(\theta)
+
\log
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right)
\right\}.
$$

The term \(\log p_k(\theta)\) increases with \(\theta\). Therefore, after a correct answer, it adds more support to higher ability values.

The shift is especially strong when the item is highly difficult because a correct response to such an item is unlikely for a low-ability user. It therefore provides strong evidence that the user's ability may be higher than previously believed.

In summary:

$$
\text{correct answer to a difficult item}
\quad\Longrightarrow\quad
\text{posterior shifts toward higher } \theta.
$$

## Question 5: Tracking Certainty and Sharpness

The discrimination parameter \(a_k\) controls how steeply the probability of a correct response changes around the item difficulty \(b_k\).

For item \(k\), the response probability is

$$
p_k(\theta)
=
\frac{1}{1+\exp\left[-a_k(\theta-b_k)\right]}.
$$

The derivative of this probability with respect to ability is

$$
\frac{dp_k(\theta)}{d\theta}
=
a_k p_k(\theta)\left[1-p_k(\theta)\right].
$$

The slope is largest when

$$
p_k(\theta)=0.5,
$$

which occurs at

$$
\theta=b_k.
$$

At this point, the maximum slope is

$$
\left.
\frac{dp_k(\theta)}{d\theta}
\right|_{\theta=b_k}
=
\frac{a_k}{4}.
$$

Therefore, a larger value of \(a_k\) produces a steeper item response curve and a smaller value produces a flatter curve.

### When \(a_k\) is very large

When \(a_k\) is large, the likelihood changes rapidly around \(b_k\). A correct or incorrect response strongly separates ability values below and above the item difficulty.

The new posterior is

$$
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k)}
\right)
\propto
[p_k(\theta)]^{y_k}
[1-p_k(\theta)]^{1-y_k}
f_{\Theta\mid Y}
\left(
\theta\mid\mathbf{y}^{(k-1)}
\right).
$$

Because the likelihood is highly concentrated, it removes more probability mass from ability values that are inconsistent with the observed response. Consequently, the posterior generally becomes narrower and sharper, and its variance decreases more strongly.

A highly discriminating item therefore provides more information about the user's ability, especially when \(b_k\) is close to the user's current estimated ability.

### When \(a_k\) is very small

When \(a_k\) is small, the item response curve is nearly flat. The probability of a correct response changes only slightly as \(\theta\) changes.

Therefore, the observed response provides weak evidence for distinguishing between different ability values. Multiplying the previous posterior by this relatively flat likelihood causes only a small change in the posterior.

As a result, the posterior remains relatively broad, its variance changes very little, and uncertainty about the user's ability remains high.

### Summary

$$
\text{Large } a_k
\quad\Longrightarrow\quad
\text{steep likelihood, more information, sharper posterior, smaller variance}.
$$

$$
\text{Small } a_k
\quad\Longrightarrow\quad
\text{flat likelihood, less information, broader posterior, little variance reduction}.
$$

However, discrimination is most useful when the item difficulty \(b_k\) is near the user's likely ability. Even an item with a large \(a_k\) may provide limited information if its difficulty is far from the range of plausible ability values.
