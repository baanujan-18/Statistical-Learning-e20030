# 3)  Measurability of $E[X \mid Y]$


By definition, the conditional expectation $E[X \mid Y]$ is a random variable that is $\sigma(Y)$-measurable.For a random variable to be measurable with respect to a specific $\sigma$-algebra, its value cannot change within the finest non-empty sets (atoms) of that $\sigma$-algebra.

The atoms of $\sigma(Y)$ are exactly $A_0 = \{\omega_1, \omega_2\}$ and $A_1 = \{\omega_3, \omega_4\}$.If $E[X \mid Y]$ took different values for $\omega_1$ and $\omega_2$, we could construct a preimage mapping to a single value that isolates $\omega_1$ from $\omega_2$.

This preimage would not belong to $\sigma(Y)$, violating measurability. Therefore, $E[X \mid Y]$ must be constant on $\{\omega_1, \omega_2\}$ and constant on $\{\omega_3, \omega_4\}$.

# 4) Explicit Computation of $E[X \mid Y]$


Since $E[X \mid Y]$ is constant on the atoms of $\sigma(Y)$, its values are given by the conditional expectations over those events:

On the set $\{Y = 0\} = \{\omega_1, \omega_2\}$:$$E[X \mid Y = 0] = \frac{E[X \cdot \mathbb{1}_{\{Y=0\}}]}{P(Y=0)} = \frac{X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\})}{P(\{\omega_1\}) + P(\{\omega_2\})}$$$$E[X \mid Y = 0] = \frac{0(0.50) + 1(0.10)}{0.50 + 0.10} = \frac{0.10}{0.60} = \frac{1}{6}
$$On the set $\{Y = 1\} = \{\omega_3, \omega_4\}$:$$E[X \mid Y = 1] = \frac{E[X \cdot \mathbb{1}_{\{Y=1\}}]}{P(Y=1)} = \frac{X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\})}{P(\{\omega_3\}) + P(\{\omega_4\})}$$$$E[X \mid Y = 1] = \frac{0(0.20) + 1(0.20)}{0.20 + 0.20} = \frac{0.20}{0.40} = \frac{1}{2}$$Combining these using indicator functions gives:$$E[X \mid Y] = \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}$$

# 8) Calculating the Mean Square Error (MSE)

The Mean Squared Error (MSE) when using $E[X \mid Y]$ as the estimator for $X$ is defined as:
$$E\left[(X - E[X \mid Y])^2\right]$$

The expanded formula across the sample space is:
$$\text{MSE} = \sum_{i=1}^4 \left( X(\omega_i) - E[X \mid Y](\omega_i) \right)^2 P(\{\omega_i\})$$

---

### **Component Calculations:**

* **For $\omega_1$:**
  $$\left(0 - \frac{1}{6}\right)^2 \cdot 0.50 = \frac{1}{36} \cdot \frac{1}{2} = \frac{1}{72}$$

* **For $\omega_2$:**
  $$\left(1 - \frac{1}{6}\right)^2 \cdot 0.10 = \frac{25}{36} \cdot \frac{1}{10} = \frac{2.5}{36} = \frac{5}{72}$$

* **For $\omega_3$:**
  $$\left(0 - \frac{1}{2}\right)^2 \cdot 0.20 = \frac{1}{4} \cdot \frac{1}{5} = \frac{1}{20} = \frac{3.6}{72}$$

* **For $\omega_4$:**
  $$\left(1 - \frac{1}{2}\right)^2 \cdot 0.20 = \frac{1}{4} \cdot \frac{1}{5} = \frac{1}{20} = \frac{3.6}{72}$$

---

### **Total MSE:**
$$\text{MSE} = \frac{1 + 5 + 3.6 + 3.6}{72} = \frac{13.2}{72} = \frac{11}{60} \approx 0.1833$$

# 9) Ignoring $Y$ vs. Law of Total Variance

If the bank ignores $Y$ and uses the global constant estimate $E[X] = 0.30$, the new error is simply the global variance of $X$:
$$\text{MSE}_{\text{constant}} = \text{Var}(X) = E[X^2] - (E[X])^2$$

Since $X$ is a binary random variable ($X^2 = X$), we have $E[X^2] = E[X] = 0.30$.
$$\text{Var}(X) = 0.30 - (0.30)^2 = 0.30 - 0.09 = 0.21$$

---

### **Connection to the Law of Total Variance**

The Law of Total Variance states:
$$\text{Var}(X) = E[\text{Var}(X \mid Y)] + \text{Var}(E[X \mid Y])$$

Where:
* **$E[\text{Var}(X \mid Y)]$**: The remaining internal system uncertainty after observing $Y$. This corresponds exactly to our calculated minimum $\text{MSE}$ from Question 8:
  $$\text{MSE} = \frac{11}{60} \approx 0.1833$$
* **$\text{Var}(E[X \mid Y])$**: The variance explained by knowing the risk profile $Y$.

---

### **Verifying the Component: $\text{Var}(E[X \mid Y])$**

First, let's compute the second moment of the conditional expectation:
$$E\left[(E[X \mid Y])^2\right] = \left(\frac{1}{6}\right)^2 (0.60) + \left(\frac{1}{2}\right)^2 (0.40) = \frac{1}{36}\left(\frac{3}{5}\right) + \frac{1}{4}\left(\frac{2}{5}\right) = \frac{1}{60} + \frac{6}{60} = \frac{7}{60}$$

Now, computing the variance:
$$\text{Var}(E[X \mid Y]) = \frac{7}{60} - (0.30)^2 = \frac{7}{60} - \frac{9}{100} = \frac{35 - 27}{300} = \frac{8}{300} = \frac{2}{75} \approx 0.0267$$

---

### **Checking the Identity Alignment**

Let's sum our two components to check if they perfectly reconstruct $\text{Var}(X)$:
$$\text{Var}(X) = \frac{11}{60} + \frac{2}{75} = \frac{55 + 8}{300} = \frac{63}{300} = \frac{21}{100} = 0.21$$

> **Conclusion:** The identity holds perfectly. Ignoring the risk class $Y$ increases the prediction error from $0.1833$ to $0.2100$. Knowing $Y$ effectively eliminates a meaningful chunk of system variance equal to $\frac{2}{75} \approx 0.0267$.

# 10) Information Theoretic Analysis (Log Base 2)

---

### **a) Entropy $H(X)$**
The global distribution of $X$ is $P(X=1) = 0.30$ and $P(X=0) = 0.70$.

$$H(X) = -0.70 \log_2(0.70) - 0.30 \log_2(0.30) \approx 0.3651 + 0.5162 = 0.8813 \text{ bits}$$

---

### **b) Conditional Entropy $H(X \mid Y)$**
The conditional entropy is weighted across the distribution of $Y$:
$$H(X \mid Y) = P(Y=0)H(X \mid Y=0) + P(Y=1)H(X \mid Y=1)$$

* **For $H(X \mid Y=0)$:** (Using the distribution $(\frac{5}{6}, \frac{1}{6})$)
  $$H(X \mid Y=0) = -\frac{5}{6}\log_2\left(\frac{5}{6}\right) - \frac{1}{6}\log_2\left(\frac{1}{6}\right) \approx 0.6500 \text{ bits}$$

* **For $H(X \mid Y=1)$:** (Using the distribution $(\frac{1}{2}, \frac{1}{2})$)
  $$H(X \mid Y=1) = -\frac{1}{2}\log_2\left(\frac{1}{2}\right) - \frac{1}{2}\log_2\left(\frac{1}{2}\right) = 1.0000 \text{ bit}$$

Combining these yields:
$$H(X \mid Y) = 0.60(0.6500) + 0.40(1.0000) = 0.3900 + 0.4000 = 0.7900 \text{ bits}$$

---

### **c) Mutual Information $I(X; Y)$**
$$I(X; Y) = H(X) - H(X \mid Y) = 0.8813 - 0.7900 = 0.0913 \text{ bits}$$

---

### **d) Percentage of Uncertainty Removed**
$$\text{Uncertainty Removed \%} = \frac{I(X; Y)}{H(X)} \times 100\% = \frac{0.0913}{0.8813} \times 100\% \approx 10.36\%$$

> **Conclusion:** Knowing the customer's risk class $Y$ directly resolves roughly **10.36%** of the bank's initial uncertainty regarding whether a given transaction is fraudulent.

# 2) Algebra generated by Y , sigma -(y)

___

To find the $\sigma$-algebra generated by the random variable $Y$, we identify the preimages of all possible values in the range of $Y$. In this case, $Y$ represents the customer risk class:

* **Low-risk:** $\{Y = 0\} = Y^{-1}(\{0\}) = \{\omega_1, \omega_2\}$
* **High-risk:** $\{Y = 1\} = Y^{-1}(\{1\}) = \{\omega_3, \omega_4\}$

The $\sigma$-algebra $\sigma(Y)$ is the collection of all sets that can be formed by taking countable unions, intersections, and complements of these "foundational blocks" (the atoms of the $\sigma$-algebra).

$$\sigma(Y) = \left\{ \emptyset, \{\omega_1, \omega_2\}, \{\omega_3, \omega_4\}, \Omega \right\}$$

___

#3. Measurability of $E[X \mid Y]$

By definition, the conditional expectation $E[X \mid Y]$ is a random variable that is $\sigma(Y)$-measurable.

For a random variable to be measurable with respect to a specific $\sigma$-algebra, its value must be constant within the finest non-empty sets (atoms) of that $\sigma$-algebra. The atoms of $\sigma(Y)$ are exactly:
* $A_0 = \{\omega_1, \omega_2\}$
* $A_1 = \{\omega_3, \omega_4\}$

---

### **Why it must be constant on these sets:**

If $E[X \mid Y]$ were to take different values for $\omega_1$ and $\omega_2$, we could define a Borel set $B$ such that only one of those values (e.g., the value at $\omega_1$) is contained in $B$. The preimage $(E[X \mid Y])^{-1}(B)$ would then contain $\omega_1$ but not $\omega_2$.

However, since $\omega_1$ and $\omega_2$ always appear together in every set belonging to $\sigma(Y)$, any set containing $\omega_1$ but not $\omega_2$ **cannot** belong to $\sigma(Y)$. This would directly violate the requirement of measurability.

Therefore:
* $E[X \mid Y]$ must be constant on $\{\omega_1, \omega_2\}$.
* $E[X \mid Y]$ must be constant on $\{\omega_3, \omega_4\}$.
