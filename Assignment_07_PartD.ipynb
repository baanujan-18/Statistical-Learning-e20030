# Bayesian Estimations for Structural Health Monitoring via Bounded Grid Updates

Let $\Theta=\theta$ represent the remaining structural stiffness efficiency factor, where

$$
\theta\in(0,1].
$$

A value of $\theta=1$ represents a completely healthy structure, while values approaching zero represent severe structural degradation.

Let $K_{\mathrm{nominal}}$ be the nominal stiffness of the completely healthy structure.

At inspection step $k$, the measured stiffness is modeled as

$$
Y_k
=
\theta K_{\mathrm{nominal}}\exp(\varepsilon_k),
$$

where

$$
\varepsilon_k\sim N(0,\sigma^2).
$$

The multiplicative log-normal noise guarantees that the measured stiffness is always positive.
