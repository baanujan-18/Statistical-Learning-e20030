### 4. Duplicate and Outlier Handling

The system will include a function to remove duplicate records from the dataset. Duplicate records can affect the quality of analysis, so removing them is an important preprocessing step.

Outlier detection will be performed using the **Interquartile Range method**.

The IQR method identifies abnormal values using the following equations:

$$
IQR = Q_3 - Q_1
$$

$$
\text{Lower Bound} = Q_1 - 1.5 \times IQR
$$

$$
\text{Upper Bound} = Q_3 + 1.5 \times IQR
$$

Values below the lower bound or above the upper bound will be identified as possible outliers. The system will allow users to detect and optionally remove these outliers.

# Step 12: Add Automatic Data-Type Correction

````python
# ==========================================
# Step 12: Add Automatic Data-Type Correction
# ==========================================

class DataInspector:

    def __init__(self):
        self.df = None  # No dataset is stored when the object is first created

    def load_from_url(self, url):
        self.df = pd.read_csv(url)  # Reads the CSV dataset from the given URL
        print("Dataset loaded successfully.")  # Displays a confirmation message
        return self.df  # Returns the loaded DataFrame

    def clean_invalid_values(self):

        # Checks whether a dataset has been loaded
        if self.df is None:
            print("No dataset is loaded. Please load a dataset first.")
            return None

        # Stores common invalid values found in real-world datasets
        invalid_values = [
            "?",
            "n/a",
            "N/A",
            "NA",
            "NULL",
            "null",
            "",
            " "
        ]

        # Replaces invalid values with proper missing values
        self.df.replace(invalid_values, np.nan, inplace=True)

        print("Invalid values converted to NaN successfully.")

        return self.df

    def correct_data_types(self):

        # Checks whether a dataset has been loaded
        if self.df is None:
            print("No dataset is loaded. Please load a dataset first.")
            return None

        # Goes through each column in the dataset one by one
        for column in self.df.columns:

            # Attempts to convert the current column into numerical values
            converted_column = pd.to_numeric(self.df[column], errors="coerce")

            # Counts the number of available values in the original column
            original_non_null_count = self.df[column].notna().sum()

            # Counts the number of available values after attempting numerical conversion
            converted_non_null_count = converted_column.notna().sum()

            # Converts the column only if all available values can be safely converted
            if original_non_null_count == converted_non_null_count:
                self.df[column] = converted_column

        print("Suitable columns converted into numerical data types successfully.")

        return self.df
````

# Step 13: Test Automatic Data-Type Correction

````python
# ==========================================
# Step 13: Test Automatic Data-Type Correction
# ==========================================

inspector = DataInspector()  # Creates a new object using the updated class

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"  # Stores the dataset URL

df = inspector.load_from_url(url)  # Loads the dataset

df = inspector.clean_invalid_values()  # Cleans invalid values

df = inspector.correct_data_types()  # Converts suitable columns into numerical formats

df.dtypes  # Displays the data type of each column
````
