## Main Functionalities to be Developed

Examples of invalid values include:

- `"null"`

Suitable columns will also be automatically converted into numerical data types wherever possible. This helps to reduce manual preprocessing work and improves the reliability of later analysis.

### 4. Duplicate and Outlier Handling

The system will include a function to remove duplicate records from the dataset. Duplicate records can affect the quality of analysis, so removing them is an important preprocessing step.

Outlier detection will be performed using the **Interquartile Range method**.

The IQR method identifies abnormal values using the following equations:

$$
IQR = Q_3 - Q_1
$$

$$
\text{Lower Bound} = Q_1 - 1.5 \times IQR
$$

$$
\text{Upper Bound} = Q_3 + 1.5 \times IQR
$$

Values below the lower bound or above the upper bound will be identified as possible outliers. The system will allow users to detect and optionally remove these outliers.

---

### 5. Feature Engineering Preparation

The project will include normalization and encoding methods to prepare the dataset for machine learning or further analysis.

Numerical columns will be normalized using:

- Min-Max Scaling
- Standard Scaling
- Robust Scaling

Categorical columns will be encoded using:

- One-Hot Encoding
- Ordinal Encoding
- Uniform Encoding

After applying scaling and encoding, the processed numerical and categorical data will be combined into a final cleaned DataFrame.

### 7. Association Heatmap

The final stage of the project is to create a unified association heatmap. This heatmap will be used to analyze relationships between numerical and categorical variables.

Different statistical techniques will be used depending on the variable types:

$$
\text{Numerical vs Numerical} \rightarrow \text{Pearson Correlation}
$$

$$
\text{Categorical vs Categorical} \rightarrow \text{Cramer's V}
$$

$$
\text{Numerical vs Categorical} \rightarrow \text{Point-Biserial Correlation or Eta Correlation}
$$

This heatmap will provide deeper insight into the relationships between different features in the dataset.
