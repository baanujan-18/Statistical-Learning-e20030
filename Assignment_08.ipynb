# Question 1: Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$.

The energy eigenvalues are

$$
E_j=\hbar\omega\left(j+\frac{1}{2}\right),
\qquad j=0,1,2,\ldots
$$

Define

$$
\beta=\frac{1}{k_B T}
$$

and

$$
x=\beta\hbar\omega=\frac{\hbar\omega}{k_B T}.
$$

---

## 1. Partition Function

The canonical partition function is

$$
Z(T)=\sum_{j=0}^{\infty}e^{-\beta E_j}.
$$

Substituting the energy levels,

$$
Z(T)
=
\sum_{j=0}^{\infty}
\exp\left[
-\beta\hbar\omega\left(j+\frac{1}{2}\right)
\right].
$$

Separate the term that does not depend on $j$:

$$
Z(T)
=
e^{-\beta\hbar\omega/2}
\sum_{j=0}^{\infty}
\left(e^{-\beta\hbar\omega}\right)^j.
$$

This is a geometric series with common ratio

$$
r=e^{-\beta\hbar\omega}.
$$

Since $T>0$, we have

$$
0<e^{-\beta\hbar\omega}<1,
$$

so the geometric series converges.

Using

$$
\sum_{j=0}^{\infty}r^j=\frac{1}{1-r},
$$

we obtain

$$
\boxed{
Z(T)=
\frac{
e^{-\hbar\omega/(2k_B T)}
}{
1-e^{-\hbar\omega/(k_B T)}
}
}
$$

Using

$$
2\sinh(y)=e^y-e^{-y},
$$

the partition function may also be written as

$$
\boxed{
Z(T)=
\frac{1}{
2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)
}
}.
$$

---

## 2. Helmholtz Free Energy

The Helmholtz free energy is defined by

$$
F(T)=-k_B T\ln Z(T).
$$

From the partition function,

$$
\ln Z
=
-\frac{\hbar\omega}{2k_B T}
-
\ln\left(
1-e^{-\hbar\omega/(k_B T)}
\right).
$$

Therefore,

$$
\boxed{
F(T)
=
\frac{\hbar\omega}{2}
+
k_B T
\ln\left(
1-e^{-\hbar\omega/(k_B T)}
\right)
}
$$

An equivalent hyperbolic form is

$$
\boxed{
F(T)
=
k_B T
\ln\left[
2\sinh\left(
\frac{\hbar\omega}{2k_B T}
\right)
\right].
}
$$

---

## 3. Internal Energy

The internal energy is

$$
U(T)
=
-\frac{\partial\ln Z}{\partial\beta}.
$$

Using

$$
\ln Z
=
-\frac{\beta\hbar\omega}{2}
-
\ln\left(
1-e^{-\beta\hbar\omega}
\right),
$$

differentiate with respect to $\beta$:

$$
\frac{\partial\ln Z}{\partial\beta}
=
-\frac{\hbar\omega}{2}
-
\frac{
\hbar\omega e^{-\beta\hbar\omega}
}{
1-e^{-\beta\hbar\omega}
}.
$$

Therefore,

$$
U(T)
=
\frac{\hbar\omega}{2}
+
\frac{
\hbar\omega e^{-\beta\hbar\omega}
}{
1-e^{-\beta\hbar\omega}
}.
$$

Multiplying the second fraction by
$e^{\beta\hbar\omega}/e^{\beta\hbar\omega}$ gives

$$
\boxed{
U(T)
=
\frac{\hbar\omega}{2}
+
\frac{
\hbar\omega
}{
e^{\hbar\omega/(k_B T)}-1
}
}
$$

The first term is the zero-point energy. The second term is the thermal energy.

The internal energy may also be written as

$$
\boxed{
U(T)
=
\frac{\hbar\omega}{2}
\operatorname{coth}\left(
\frac{\hbar\omega}{2k_B T}
\right).
}
$$

---

## 4. Entropy

The entropy is

$$
S(T)
=
-\left(
\frac{\partial F}{\partial T}
\right).
$$

Using

$$
x=\frac{\hbar\omega}{k_B T},
$$

the free energy becomes

$$
F(T)
=
\frac{\hbar\omega}{2}
+
k_B T\ln(1-e^{-x}).
$$

Differentiating with respect to $T$ gives

$$
\boxed{
S(T)
=
k_B
\left[
\frac{x}{e^x-1}
-
\ln(1-e^{-x})
\right].
}
$$

Substituting back

$$
x=\frac{\hbar\omega}{k_B T},
$$

we obtain

$$
\boxed{
S(T)
=
k_B
\left[
\frac{
\hbar\omega/(k_B T)
}{
e^{\hbar\omega/(k_B T)}-1
}
-
\ln\left(
1-e^{-\hbar\omega/(k_B T)}
\right)
\right].
}
$$

---

## 5. Verification of the Relation $U=F+TS$

Using

$$
F
=
\frac{\hbar\omega}{2}
+
k_B T\ln(1-e^{-x})
$$

and

$$
S
=
k_B
\left[
\frac{x}{e^x-1}
-
\ln(1-e^{-x})
\right],
$$

we have

$$
F+TS
=
\frac{\hbar\omega}{2}
+
k_B T\ln(1-e^{-x})
+
k_B T
\left[
\frac{x}{e^x-1}
-
\ln(1-e^{-x})
\right].
$$

The logarithmic terms cancel:

$$
F+TS
=
\frac{\hbar\omega}{2}
+
\frac{k_BTx}{e^x-1}.
$$

Since

$$
k_BTx=\hbar\omega,
$$

we obtain

$$
F+TS
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}{e^x-1}.
$$

Therefore,

$$
\boxed{U=F+TS}.
$$

---

## 6. Heat Capacity

The constant-volume heat capacity is

$$
C_V
=
\frac{\partial U}{\partial T}.
$$

The internal energy is

$$
U
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}{e^x-1},
$$

where

$$
x=\frac{\hbar\omega}{k_B T}.
$$

Since

$$
\frac{dx}{dT}
=
-\frac{x}{T},
$$

differentiating gives

$$
\boxed{
C_V
=
k_Bx^2
\frac{e^x}{(e^x-1)^2}.
}
$$

Substituting

$$
x=\frac{\hbar\omega}{k_B T},
$$

we get

$$
\boxed{
C_V
=
k_B
\left(
\frac{\hbar\omega}{k_B T}
\right)^2
\frac{
e^{\hbar\omega/(k_B T)}
}{
\left(
e^{\hbar\omega/(k_B T)}-1
\right)^2
}.
}
$$

---

## 7. Hyperbolic Form of the Heat Capacity

Using the identity

$$
\sinh\left(\frac{x}{2}\right)
=
\frac{e^{x/2}-e^{-x/2}}{2},
$$

we have

$$
e^x-1
=
2e^{x/2}\sinh\left(\frac{x}{2}\right).
$$

Squaring both sides,

$$
(e^x-1)^2
=
4e^x\sinh^2\left(\frac{x}{2}\right).
$$

Therefore,

$$
\frac{e^x}{(e^x-1)^2}
=
\frac{1}{
4\sinh^2(x/2)
}.
$$

Substituting this into the heat-capacity expression gives

$$
C_V
=
k_B
\frac{x^2}{
4\sinh^2(x/2)
}.
$$

Hence,

$$
\boxed{
C_V
=
k_B
\left(
\frac{\hbar\omega}{2k_B T}
\right)^2
\frac{1}{
\sinh^2\left(
\frac{\hbar\omega}{2k_B T}
\right)
}.
}
$$

---

## 8. High-Temperature Limit

At high temperature,

$$
k_B T\gg\hbar\omega.
$$

Therefore,

$$
x=\frac{\hbar\omega}{k_B T}\ll1.
$$

For small $x$,

$$
\sinh\left(\frac{x}{2}\right)
\approx
\frac{x}{2}.
$$

Thus,

$$
C_V
\approx
k_B
\left(\frac{x}{2}\right)^2
\frac{1}{(x/2)^2}.
$$

Therefore,

$$
\boxed{
C_V\longrightarrow k_B
}
$$

as $T\longrightarrow\infty$.

This agrees with the classical equipartition theorem. A one-dimensional harmonic oscillator has two quadratic energy terms:

1. Kinetic energy  
2. Potential energy  

Each quadratic term contributes

$$
\frac{1}{2}k_B T
$$

to the average energy. Therefore,

$$
U_{\text{classical}}=k_B T
$$

and

$$
\boxed{
C_V=\frac{\partial U}{\partial T}=k_B.
}
$$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# We use dimensionless units:
# k_B = 1
# hbar * omega = 1

k_B = 1.0
hbar_omega = 1.0


def oscillator_properties(T):
    """
    Calculate the thermodynamic properties of a
    quantum harmonic oscillator.

    Parameters
    ----------
    T : float or numpy array
        Temperature. T must be positive.

    Returns
    -------
    Z, F, U, S, C_V
    """

    T = np.asarray(T, dtype=float)

    if np.any(T <= 0):
        raise ValueError("Temperature must be greater than zero.")

    x = hbar_omega / (k_B * T)

    # 1 - exp(-x), calculated in a numerically stable way
    one_minus_exp_minus_x = -np.expm1(-x)

    # Partition function
    Z = np.exp(-x / 2) / one_minus_exp_minus_x

    # Helmholtz free energy
    F = (
        hbar_omega / 2
        + k_B * T * np.log(one_minus_exp_minus_x)
    )

    # Internal energy
    U = (
        hbar_omega / 2
        + hbar_omega / np.expm1(x)
    )

    # Entropy
    S = k_B * (
        x / np.expm1(x)
        - np.log(one_minus_exp_minus_x)
    )

    # Heat capacity
    C_V = k_B * (
        (x / 2) / np.sinh(x / 2)
    )**2

    return Z, F, U, S, C_V
```

```python
temperatures = np.array([0.2, 0.5, 1.0, 2.0, 5.0])

Z, F, U, S, C_V = oscillator_properties(temperatures)

verification = pd.DataFrame({
    "Temperature T": temperatures,
    "Partition function Z": Z,
    "Free energy F": F,
    "Internal energy U": U,
    "Entropy S": S,
    "F + T*S": F + temperatures * S,
    "Heat capacity C_V": C_V
})

verification["Error in U = F + T*S"] = np.abs(
    verification["Internal energy U"]
    - verification["F + T*S"]
)

verification
```

```python
T_values = np.linspace(0.05, 5.0, 500)

Z, F, U, S, C_V = oscillator_properties(T_values)

plt.figure(figsize=(8, 5))

plt.plot(T_values, C_V / k_B)

plt.axhline(
    y=1,
    linestyle="--",
    label=r"Classical limit: $C_V/k_B=1$"
)

plt.xlabel(
    r"Dimensionless temperature $k_B T/(\hbar\omega)$"
)

plt.ylabel(r"$C_V/k_B$")

plt.title(
    "Heat Capacity of a Quantum Harmonic Oscillator"
)

plt.grid(True)
plt.legend()
plt.show()
```

# Question 2: Maximum-Entropy Distribution under Mean and Variance Constraints

Let $X$ be a continuous random variable defined on

$$
-\infty < X < \infty.
$$

The given constraints are

$$
E[X]=\mu
$$

and

$$
E[(X-\mu)^2]=\sigma^2.
$$

We want to find the probability density function $p(x)$ that maximizes the differential entropy

$$
H[p]
=
-\int_{-\infty}^{\infty}
p(x)\ln p(x)\,dx.
$$

The density must satisfy the following three constraints:

### Normalization constraint

$$
\int_{-\infty}^{\infty}p(x)\,dx=1.
$$

### Mean constraint

$$
\int_{-\infty}^{\infty}x\,p(x)\,dx=\mu.
$$

### Variance constraint

$$
\int_{-\infty}^{\infty}
(x-\mu)^2p(x)\,dx
=
\sigma^2.
$$

---

## 1. Constructing the Lagrangian

Introduce the Lagrange multipliers

$$
\lambda_0,\lambda_1,\lambda_2.
$$

The constrained entropy functional is

$$
\begin{aligned}
\mathcal{L}[p]
=&
-\int_{-\infty}^{\infty}
p(x)\ln p(x)\,dx
\\
&-\lambda_0
\left(
\int_{-\infty}^{\infty}p(x)\,dx-1
\right)
\\
&-\lambda_1
\left(
\int_{-\infty}^{\infty}x\,p(x)\,dx-\mu
\right)
\\
&-\lambda_2
\left(
\int_{-\infty}^{\infty}
(x-\mu)^2p(x)\,dx-\sigma^2
\right).
\end{aligned}
$$

To find the maximizing density, take the functional derivative with respect to $p(x)$:

$$
\frac{\delta\mathcal{L}}{\delta p(x)}
=
-\ln p(x)-1
-\lambda_0
-\lambda_1x
-\lambda_2(x-\mu)^2.
$$

At the optimum,

$$
\frac{\delta\mathcal{L}}{\delta p(x)}=0.
$$

Therefore,

$$
-\ln p(x)-1-\lambda_0-\lambda_1x
-\lambda_2(x-\mu)^2=0.
$$

Solving for $\ln p(x)$ gives

$$
\ln p(x)
=
-1-\lambda_0-\lambda_1x
-\lambda_2(x-\mu)^2.
$$

Exponentiating both sides,

$$
p(x)
=
\exp(-1-\lambda_0)
\exp\left[
-\lambda_1x-\lambda_2(x-\mu)^2
\right].
$$

Let

$$
A=\exp(-1-\lambda_0).
$$

Then

$$
p(x)
=
A
\exp\left[
-\lambda_1x-\lambda_2(x-\mu)^2
\right].
$$

---

## 2. Applying the Mean Constraint

The distribution must have mean $\mu$.

Because the variance term is centered at $\mu$, the density must be symmetric around $\mu$.

Therefore, the additional linear term must vanish:

$$
\lambda_1=0.
$$

Hence,

$$
p(x)
=
A\exp\left[-\lambda_2(x-\mu)^2\right].
$$

---

## 3. Finding the Normalization Constant

The normalization constraint requires

$$
\int_{-\infty}^{\infty}
p(x)\,dx=1.
$$

Therefore,

$$
A
\int_{-\infty}^{\infty}
\exp\left[-\lambda_2(x-\mu)^2\right]dx
=
1.
$$

Using the Gaussian integral

$$
\int_{-\infty}^{\infty}
e^{-a(x-\mu)^2}dx
=
\sqrt{\frac{\pi}{a}},
$$

we obtain

$$
A\sqrt{\frac{\pi}{\lambda_2}}=1.
$$

Thus,

$$
A=\sqrt{\frac{\lambda_2}{\pi}}.
$$

---

## 4. Applying the Variance Constraint

The variance must satisfy

$$
E[(X-\mu)^2]=\sigma^2.
$$

For a density of the form

$$
p(x)
=
\sqrt{\frac{\lambda_2}{\pi}}
\exp\left[-\lambda_2(x-\mu)^2\right],
$$

the variance is

$$
\frac{1}{2\lambda_2}.
$$

Therefore,

$$
\frac{1}{2\lambda_2}
=
\sigma^2.
$$

Solving for $\lambda_2$ gives

$$
\lambda_2
=
\frac{1}{2\sigma^2}.
$$

The normalization constant becomes

$$
A
=
\sqrt{
\frac{1}{2\pi\sigma^2}
}.
$$

Therefore,

$$
A
=
\frac{1}{\sqrt{2\pi\sigma^2}}.
$$

---

## 5. Final Maximum-Entropy Distribution

Substituting the values of $A$ and $\lambda_2$, we obtain

$$
\boxed{
p(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left[
-\frac{(x-\mu)^2}{2\sigma^2}
\right]
}
$$

This is the probability density function of a normal distribution.

Therefore,

$$
\boxed{
X\sim\mathcal{N}(\mu,\sigma^2)
}
$$

is the distribution that maximizes entropy subject to fixed mean and variance.

---

## 6. Maximum Entropy Value

For a Gaussian random variable,

$$
X\sim\mathcal{N}(\mu,\sigma^2),
$$

the differential entropy is

$$
\boxed{
H(X)
=
\frac{1}{2}
\ln\left(2\pi e\sigma^2\right)
}.
$$

The entropy does not depend on the mean $\mu$.

It depends only on the variance $\sigma^2$.

A larger variance means that the distribution is more spread out, resulting in greater uncertainty and higher entropy.

---

## 7. Why This Is the Global Maximum

The entropy functional

$$
H[p]
=
-\int p(x)\ln p(x)\,dx
$$

is a concave function of $p(x)$.

The normalization, mean and variance constraints are linear constraints.

Therefore, the stationary solution obtained using the Lagrange multiplier method is the unique global maximum.

Hence, among all continuous probability distributions with mean $\mu$ and variance $\sigma^2$, the Gaussian distribution has the highest entropy.

```python
import numpy as np
import matplotlib.pyplot as plt

# Choose values for the mean and standard deviation
mu = 0.0
sigma = 1.0

# Generate x-values
x = np.linspace(
    mu - 4 * sigma,
    mu + 4 * sigma,
    500
)

# Maximum-entropy Gaussian density
p = (
    1 / np.sqrt(2 * np.pi * sigma**2)
    * np.exp(
        -(x - mu)**2 / (2 * sigma**2)
    )
)

plt.figure(figsize=(8, 5))

plt.plot(
    x,
    p,
    label=r"$\mathcal{N}(\mu,\sigma^2)$"
)

plt.xlabel("x")
plt.ylabel("p(x)")

plt.title(
    "Maximum-Entropy Distribution "
    "with Fixed Mean and Variance"
)

plt.grid(True)
plt.legend()
plt.show()
```

```python
# Numerical integration using the trapezoidal rule

normalization = np.trapz(p, x)

calculated_mean = np.trapz(
    x * p,
    x
)

calculated_variance = np.trapz(
    (x - calculated_mean)**2 * p,
    x
)

print("Required normalization =", 1)
print("Calculated normalization =", normalization)

print()

print("Required mean =", mu)
print("Calculated mean =", calculated_mean)

print()

print("Required variance =", sigma**2)
print("Calculated variance =", calculated_variance)
```

```python
# Theoretical entropy of the Gaussian distribution

gaussian_entropy = 0.5 * np.log(
    2 * np.pi * np.e * sigma**2
)

print(
    "Maximum Gaussian entropy =",
    gaussian_entropy,
    "nats"
)
```
